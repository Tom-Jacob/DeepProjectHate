# 08 — Figures for report

Displays every PNG under `report/figures/` and summarizes cached JSON metrics.
Does **not** retrain models.

```python
from pathlib import Path
import json
from IPython.display import Image, Markdown, display

_cwd = Path.cwd()
_root = _cwd.parent if _cwd.name == "notebooks" else _cwd
fig_dir = _root / "report" / "figures"
proc = _root / "data" / "processed"

for p in sorted(fig_dir.glob("*.png")):
    display(Markdown(f"### {p.name}"))
    display(Image(filename=str(p)))

for name in ("model_comparison.json", "fairness_results.json", "robustness_results.json",
             "masking_results.json", "agreement_results.json", "distill_results.json"):
    path = proc / name
    display(Markdown(f"**{name}** — exists={path.exists()}"))
```
